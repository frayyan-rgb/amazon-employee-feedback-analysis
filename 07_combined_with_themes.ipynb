```python
import pandas as pd

# Load your dataset
df = pd.read_csv('combined_sentiment_data.csv')
```

```python
theme_keywords = {
    'Burnout & Fatigue': [
        'burnout', 'burnt out', 'burned out', 'exhausted', 'exhausting', 'tired',
        'overworked', 'no break', 'no breaks', 'stress', 'drained', 'worn out',
        'double shift', 'peak season', 'physically demanding'
    ],
    'Shift Strain (Night vs Day)': [
        'night shift', 'overnight', 'graveyard', 'day shift', 'swing shift',
        'mandatory overtime', 'forced overtime', 'overtime', 'long hours',
        'shift change', 'schedule', 'weekend', '10 hour', '12 hour'
    ],
    'New-Hire Training & Support': [
        'training', 'onboarding', 'orientation', 'new hire', 'first week',
        'first day', 'no support', 'unsupported', 'confused', 'unprepared',
        'no one showed', 'figure it out', 'trained myself', 'not trained'
    ]
}
```

```python
def tag_themes(text, theme_keywords):
    matched_themes = []
    for theme, keywords in theme_keywords.items():
        if any(keyword in str(text).lower() for keyword in keywords):
            matched_themes.append(theme)
    return ', '.join(matched_themes) if matched_themes else ''
```

```python
df['Theme'] = df['cleaned_text'].apply(lambda x: tag_themes(x, theme_keywords))
```

```python
df.to_csv('combined_with_themes.csv', index=False)

```
